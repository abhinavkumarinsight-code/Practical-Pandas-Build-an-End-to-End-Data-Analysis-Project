## Section H: replace, map & apply (101–115)

* Transformed dataset values efficiently using apply, map, and replace functions.

* Enabled effective feature engineering and custom data manipulation through these methods.

```python
# Load the CSV file into a Pandas DataFrame.
import pandas as pd

df = pd.read_csv(r"C://Users//Abhinav kumar//Downloads//employee_pandas_practice.csv")
df
```

```python
# Replace "IT" with "Information Technology".

df['department'] = df['department'].replace('IT', 'Information Technology')
df['department']
```

```python
# Replace missing bonus percent with 0.

df['bonus_percent'] = df['bonus_percent'].fillna(0)
df['bonus_percent'] 
```

```python
# Replace rare education categories with "Other".

rare = df['education'].value_counts()[df['education'].value_counts() < 2].index

df['education'] = df['education'].replace(rare, 'Other')
df['education']
```

```python
# Convert gender into numeric format.

df['gender'] = df['gender'].map({'Male': 1, 'Female': 0})
df['gender'] 
```

```python
# Replace salary values greater than 1,50,000 with 1,50,000.

df['salary'] = df['salary'].clip(upper=150000)
df['salary']
```

```python
# Create age group column using apply.

df['age_group'] = df['age'].apply(lambda x: 'Young' if x < 30 else 'Middle' if x < 50 else 'Senior')
df['age_group']
```

```python
# Create performance label using rating.

df['performance'] = df['rating'].apply(lambda x: 'Low' if x < 3 else 'Average' if x < 4 else 'High')
df['performance'] 
```

```python
# Create salary slab using lambda.

df['salary_slab'] = df['salary'].apply(lambda x: 'Low' if x < 30000 else 'Medium' if x < 70000 else 'High')
df['salary_slab']
```

```python
# Replace negative values if any with 0.

num_cols = df.select_dtypes(include='number').columns

df[num_cols] = df[num_cols].clip(lower=0)
df[num_cols]
```

```python
# Create combined city-department column.

df['city_department'] = df['city'].astype(str) + '_' + df['department'].astype(str)
df['city_department']
```

```python
# Replace marital status with binary values.

df['marital_status'] = df['marital_status'].map({'Married': 1, 'Single': 0})
df['marital_status']
```

```python
# Replace project count outliers using median.

Q1 = df['projects'].quantile(0.25)
Q3 = df['projects'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

median_val = df['projects'].median()

df.loc[(df['projects'] < lower) | (df['projects'] > upper), 'projects'] = median_val
median_val
```

```python
# Replace education values using dictionary mapping.

mapping = {'Bachelors': 'UG','Masters': 'PG','PhD': 'Doctorate'}
df['education'] = df['education'].map(mapping)
df['education']
```

```python
# Apply row-wise condition on salary and experience.

df['category'] = df.apply(
    lambda x: 'High' if (x['salary'] > 70000 and x['experience'] > 5) else 'Low',
    axis=1)

df['category']
```

```python
# Create salary after bonus column.

df['salary_after_bonus'] = df['salary'] + (df['salary'] * df['bonus_percent'] / 100)
df['salary_after_bonus']
```

## Section I: GroupBy & Aggregation (116–135)

* Performed grouped data summaries to uncover deeper patterns and trends.

* Executed industry-standard analytics practices to support robust data reporting.

```python
# Calculate average salary per department.

df.groupby('department', observed=True)['salary'].mean()
```

```python
# Calculate total projects per department.

df.groupby('department', observed=True)['projects'].sum()
```

```python
# Calculate average rating per city.

df.groupby('city')['rating'].mean()
```

```python
# Count employees per department.

df.groupby('department', observed=True)['id'].count()
```

```python
# Calculate department-wise max salary.

df.groupby('department', observed=True)['salary'].max()
```

```python
# Calculate department-wise min experience.

df.groupby('department', observed=True)['experience'].min()
```

```python
# Calculate city-wise average bonus percent.

df.groupby('city')['bonus_percent'].mean()
```

```python
# Group by department and city together.

df.groupby(['department', 'city'], observed=True).size()
```

```python
# Find department with highest average salary.

df.groupby('department', observed=True)['salary'].mean().idxmax()
```

```python
# Find city with lowest average rating.

df.groupby('city')['rating'].mean().idxmin()
```

```python
# Group-wise missing value count.

df.groupby('department', observed=True).agg(lambda x: x.isnull().sum())
```

```python
# Group-wise salary standard deviation.

df.groupby('department', observed=True)['salary'].std()
```

```python
# Group-wise median salary.

df.groupby('department', observed=True)['salary'].median()
```

```python
# Calculate department-wise salary range.

df.groupby('department', observed=True)['salary'].agg(lambda x: x.max() - x.min())
```

```python
# Find top-paid employee per department.

df.loc[df.groupby('department', observed=True)['salary'].idxmax()]
```

```python
# Find least experienced employee per city.

df.loc[df.groupby('city')['experience'].idxmin()]
```

```python
# Calculate salary contribution percentage per city.

df['city_salary_pct'] = df['salary'] / df['salary'].sum() * 100
df['city_salary_pct']
```

```python
# Group-wise normalization of salary.

df['salary_normalized'] = df.groupby('department', observed=True)['salary'].transform(
    lambda x: (x - x.min()) / (x.max() - x.min()))
df['salary_normalized']
```

```python
# Department-wise cumulative employee count.

df['dept_cum_count'] = df.groupby('department', observed=True).cumcount() + 1
df['dept_cum_count'] 
```

```python
# Group-wise outlier detection for salary.

def detect_outliers(group):
    Q1 = group['salary'].quantile(0.25)
    Q3 = group['salary'].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return group[(group['salary'] < lower) | (group['salary'] > upper)]

df.groupby('department', observed=True, group_keys=False)\
  .apply(detect_outliers, include_groups=False)
```

## Section J: Binning using pd.cut / pd.qcut (136–145)

* Discretized continuous numerical features into categorical bins.

* Optimized analytical workflows and enhanced the effectiveness of visual data representations.

```python
# Create salary bins using pd.cut.

df['salary_bin'] = pd.cut(df['salary'], bins=3)
df['salary_bin'] 
```

```python
# Label salary bins as Low, Medium, High.

df['salary_bin'] = pd.cut(df['salary'], bins=3, labels=['Low', 'Medium', 'High'])
df['salary_bin']
```

```python
# Create age groups using fixed bins.

df['age_group'] = pd.cut(df['age'], bins=[0, 30, 50, 100], labels=['Young', 'Middle', 'Senior'])
df['age_group'] 
```

```python
# Create experience quartiles using pd.qcut.

df['exp_quartile'] = pd.qcut(df['experience'], q=4)
df['exp_quartile']
```

```python
# Compare bin sizes from cut vs qcut.

# Count from cut
cut_counts = df['salary_bin'].value_counts()

# Count from qcut
qcut_counts = pd.qcut(df['salary'], q=3).value_counts()

cut_counts, qcut_counts
```

```python
# Identify highest salary bin employees.

df[df['salary_bin'] == 'High']
```

```python
# Create project workload bins.

df['project_workload'] = pd.cut(df['projects'], bins=3, labels=['Low', 'Medium', 'High'])
df['project_workload']
```
