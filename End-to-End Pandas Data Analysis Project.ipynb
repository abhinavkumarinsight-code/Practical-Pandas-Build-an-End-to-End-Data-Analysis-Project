## Section E: Filtering & Conditional Selection (51–70)

* Filtered the data to keep only the useful details.

* This made analysis easier and helped in making better decisions.

```python
# Load the CSV file into a Pandas DataFrame.
import pandas as pd

df = pd.read_csv(r"C://Users//Abhinav kumar//Downloads//employee_pandas_practice.csv")
df
```

```python
# Filter employees whose salary is greater than 60,000.

df[df['salary'] > 60000]
```

```python
# Filter employees whose age is less than 30.

df[df['age'] < 30]
```

```python
# Filter employees from Delhi or Mumbai.

df[df['city'].isin(['Delhi', 'Mumbai'])]
```

```python
# Filter employees from IT department with salary > 70,000.

df[(df['department'] == 'IT') & (df['salary'] > 70000)]
```

```python
# Filter employees with more than 10 years of experience.

df[df['experience'] > 10]
```

```python
# Filter employees who are active.

df[df['is_active'] == True]
```

```python
# Filter employees who are inactive and married.

df[(df['is_active'] == False) & (df['marital_status'] == 'Married')]
```

```python
# Filter employees whose projects count is between 3 and 7.

df[(df['projects'] >= 3) & (df['projects'] <= 7)]
```

```python
# Filter employees whose salary is missing.

df[df['salary'].isnull()]
```

```python
# Filter employees with no missing values.

df[df.notnull().all(axis=1)]
```

```python
# Filter employees whose rating is above average.

df[df['rating'] > df['rating'].mean()]
```

```python
# Filter top 5 highest-paid employees.

df.nlargest(5, 'salary')
```

```python
# Filter bottom 5 employees by rating.

df.nsmallest(5, 'rating')
```

```python
# Filter employees whose city is not Bangalore.

df[df['city'] != 'Bangalore']
```

```python
# Filter employees whose bonus percent is greater than 15.

df[df['bonus_percent'] > 15]
```

```python
# Filter employees using query() method.

df.query("salary > 60000 and age < 30")
```

```python
# Filter employees using lambda function.

df[lambda x: (x['salary'] > 50000) & (x['age'] < 40)]
```

```python
# Filter rows where salary is above department average.

df[df['salary'] > df.groupby('department', observed=True)['salary'].transform('mean')]
```

```python
# Filter employees with maximum experience.

df[df['experience'] == df['experience'].max()]
```

```python
# Filter employees whose projects count is an outlier.

Q1 = df['projects'].quantile(0.25)
Q3 = df['projects'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df[(df['projects'] < lower) | (df['projects'] > upper)]
```

## Section F: value_counts, unique, nunique (71–85)
* Explored categorical variables using frequency counts and category distributions.
* Identified key patterns and trends across different categories in the dataset.


```python
# Find unique cities.

df['city'].unique()
```

```python
# Count number of unique departments.

df['department'].nunique()
```

```python
# Find frequency of each department.

df['department'].value_counts()
```

```python
# Find city-wise employee count.

df['city'].value_counts()
```

```python
#Find education-wise distribution.

df['education'].value_counts()
```

```python
# Calculate percentage distribution of gender.

df['gender'].value_counts(normalize=True) * 100
```

```python
# Sort department counts in ascending order.

df['department'].value_counts().sort_values(ascending=True)
```

```python
# Identify departments having more than 3 employees.

df['department'].value_counts()[df['department'].value_counts() > 3]
```

```python
# Convert value_counts output to DataFrame.

df['department'].value_counts().reset_index(name='count')
```

```python
# Find rare education categories.

df['education'].value_counts()[df['education'].value_counts() < 2]
```

```python
# Count unique cities per department.

df.groupby('department', observed=True)['city'].nunique()
```

```python
# Find most common city.

df['city'].value_counts().idxmax()
```

```python
# Find least common department.

df['department'].value_counts().idxmin()
```

```python
# Find number of unique marital statuses.

df['marital_status'].nunique()
```

```python
# Merge rare education values into "Other".

rare = df['education'].value_counts()[df['education'].value_counts() < 2].index

df['education'] = df['education'].replace(rare, 'Other')
df ['education']
```

## Section G: Sorting & Ranking (86–100)

* Used sorting and ranking techniques to analyze trends, top and bottom values, and performance indicators.
* This helps in making data comparisons clearer and easier to understand.


```python
# Sort employees by salary ascending.

df.sort_values(by='salary', ascending=True)
```

```python
# Sort employees by salary descending

df.sort_values(by='salary', ascending=False)
```

```python
# Sort employees by city and salary.

df.sort_values(by=['city', 'salary'])
```

```python
# Sort employees by experience descending.

df.sort_values(by='experience', ascending=False)
```

```python
# Sort employees placing missing salary at the end.

df.sort_values(by='salary', na_position='last')
```

```python
# Rank employees based on salary.

df['salary_rank'] = df['salary'].rank()
df['salary_rank']
```

```python
# Rank employees department-wise by salary.

df['dept_salary_rank'] = df.groupby('department', observed=True)['salary'].rank()
df['dept_salary_rank']
```

```python
# Rank employees based on experience.

df['experience_rank'] = df['experience'].rank()
df['experience_rank'] 
```

```python
# Sort departments by average salary.

df.groupby('department', observed=True)['salary'].mean().sort_values()
```

```python
# Sort employees by number of projects.

df.sort_values(by='projects')
```

```python
# Sort DataFrame by index descending.

df.sort_index(ascending=False)
```

```python
# Sort employees by salary-to-experience ratio.

df.assign(ratio=df['salary'] / df['experience']).sort_values(by='ratio', ascending=False)
```

```python
# Sort employees by bonus percent.

df.sort_values(by='bonus_percent')
```

```python
# Sort employees by frequency of city.

df.sort_values(by='city', key=lambda x: x.map(x.value_counts()))
```

```python
# Sort employees based on custom department order.

order = ['HR', 'IT', 'Finance', 'Sales']

df['department'] = pd.Categorical(df['department'], categories=order, ordered=True)

df.sort_values(by='department')
```
